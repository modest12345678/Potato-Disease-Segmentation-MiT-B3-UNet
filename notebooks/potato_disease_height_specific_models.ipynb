```python
import torch
import numpy as np
import segmentation_models_pytorch as smp
from torch.utils.data import DataLoader
from train_hybrid import PotatoDataset, get_preprocessing, CONFIG, get_validation_augmentation
from pathlib import Path

def evaluate(height):
    device = "cuda" if torch.cuda.is_available() else "cpu"

    # 1. Load the brain we just trained
    model = smp.Unet(
        encoder_name=CONFIG['ENCODER'],
        encoder_weights=None,
        classes=len(CONFIG['CLASSES'])
    )

    print(f"\n--- Loading best_hybrid_model_{height}.pth ---")
    try:
        checkpoint = torch.load(f"best_hybrid_model_{height}.pth", map_location=device, weights_only=False)
        model.load_state_dict(checkpoint["model_state_dict"])
    except FileNotFoundError:
        print(f"❌ Error: Could not find the file for {height}. Did you finish training it?")
        return

    model.to(device)
    model.eval()

    # 2. Load the hidden test pictures
    preprocessing_fn = smp.encoders.get_preprocessing_fn(CONFIG['ENCODER'], "imagenet")
    DATA_DIR = Path(f"datasets/potato_processed/{height}")

    test_dataset = PotatoDataset(
        DATA_DIR / "images/test",
        DATA_DIR / "masks/test",
        augmentation=get_validation_augmentation(),
        preprocessing=get_preprocessing(preprocessing_fn),
    )

    if len(test_dataset) == 0:
        print(f"❌ Error: No test pictures found for {height}!")
        return

    test_loader = DataLoader(test_dataset, batch_size=1, shuffle=False)
    print(f"Testing the AI on {len(test_dataset)} never-before-seen pictures...")

    correct_pixels = 0
    total_pixels = 0

    # 3. Calculate Accuracy
    with torch.no_grad():
        for img, mask in test_loader:
            img, mask = img.to(device), mask.to(device)

            logits = model(img)
            predictions = logits.argmax(dim=1)

            correct_pixels += (predictions == mask).sum().item()
            total_pixels += torch.numel(mask)

    accuracy = (correct_pixels / total_pixels) * 100
    print(f"✅ FINAL {height} MODEL ACCURACY: {accuracy:.2f}%")

# Check BOTH models automatically!
evaluate("7m")
evaluate("2m")
```
